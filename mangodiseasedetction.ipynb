```python
# ============================================================
# MANGO DISEASE DETECTION (IMPROVED CBAM + VGG16)
# FIXED FOR TF 2.20 / KERAS 3 + HIGH ACCURACY
# ALL BUGS FIXED - COMPLETE ALL-IN-ONE CODE
# ============================================================


!pip install -q tf-keras-vis seaborn


import os, zipfile
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


from google.colab import files
import tensorflow as tf


from tensorflow.keras.layers import (
    Dense, Dropout, BatchNormalization, GlobalAveragePooling2D,
    GlobalMaxPooling2D, Reshape, Multiply, Add, Activation,
    Conv2D, Concatenate, Lambda
)
from tensorflow.keras.models import Model
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img, img_to_array
from tensorflow.keras.applications.vgg16 import preprocess_input
from tensorflow.keras.applications import VGG16


from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score


from tf_keras_vis.gradcam_plus_plus import GradcamPlusPlus
from tf_keras_vis.utils.scores import CategoricalScore


print("TF:", tf.__version__)
print("GPU:", tf.config.list_physical_devices('GPU'))


# ============================================================
# STEP 1: UPLOAD DATASET
# ============================================================


uploaded = files.upload()
zip_file = list(uploaded.keys())[0]


extract_path = "/content/maize_dataset"


with zipfile.ZipFile(zip_file, 'r') as z:
    z.extractall(extract_path)


dataset_dir = None
for root, dirs, fnames in os.walk(extract_path):
    if len(dirs) >= 2:
        dataset_dir = root
        break


print("Dataset:", dataset_dir)


# ============================================================
# STEP 2: DATA AUGMENTATION (FIXED WITH SEED)
# ============================================================


SEED = 42  # FIXED: Add random seed for consistent split


train_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input,
    validation_split=0.2,
    rotation_range=30,
    zoom_range=0.25,
    width_shift_range=0.2,
    height_shift_range=0.2,
    horizontal_flip=True,
    vertical_flip=False,
    fill_mode='nearest'
)


test_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input,
    validation_split=0.2
)


IMG_SIZE = (224, 224)
BATCH = 32


train_generator = train_datagen.flow_from_directory(
    dataset_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH,
    class_mode='categorical',
    subset='training',
    shuffle=True,
    seed=SEED  # FIXED: Add seed for consistent split
)


test_generator = test_datagen.flow_from_directory(
    dataset_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH,
    class_mode='categorical',
    subset='validation',
    shuffle=False,
    seed=SEED  # FIXED: Add seed for consistent split
)


print("Classes:", train_generator.class_indices)


# ============================================================
# STEP 3: CBAM BLOCK (IMPROVED VERSION)
# ============================================================


def cbam_block(inputs, ratio=8):
    channel = inputs.shape[-1]


    # Channel attention
    avg = GlobalAveragePooling2D()(inputs)
    maxp = GlobalMaxPooling2D()(inputs)


    avg = Dense(channel // ratio, activation='relu', kernel_initializer='he_normal')(avg)
    avg = Dense(channel, kernel_initializer='he_normal')(avg)


    maxp = Dense(channel // ratio, activation='relu', kernel_initializer='he_normal')(maxp)
    maxp = Dense(channel, kernel_initializer='he_normal')(maxp)


    channel_attention = Activation('sigmoid')(Add()([avg, maxp]))
    channel_attention = Reshape((1, 1, channel))(channel_attention)


    x = Multiply()([inputs, channel_attention])


    # Spatial attention
    avg_pool = Lambda(lambda t: tf.reduce_mean(t, axis=-1, keepdims=True),
                      name='spatial_avg_pool')(x)
    max_pool = Lambda(lambda t: tf.reduce_max(t, axis=-1, keepdims=True),
                      name='spatial_max_pool')(x)


    concat = Concatenate()([avg_pool, max_pool])
    spatial = Conv2D(1, 7, padding='same', activation='sigmoid',
                     kernel_initializer='he_normal')(concat)


    return Multiply()([x, spatial])


# ============================================================
# STEP 4: BUILD MODEL (VGG16 - FIXED)
# ============================================================


base_model = VGG16(
    include_top=False,
    weights='imagenet',
    input_shape=(224, 224, 3)
)


# FIXED: Freeze base model FIRST
base_model.trainable = False


x = base_model.output
x = cbam_block(x)


# FIXED: Reduced dropout for small dataset
x = GlobalAveragePooling2D()(x)
x = Dense(512, activation='relu', kernel_initializer='he_normal')(x)
x = BatchNormalization()(x)
x = Dropout(0.3)(x)  # FIXED: Was 0.5


x = Dense(256, activation='relu', kernel_initializer='he_normal')(x)
x = BatchNormalization()(x)
x = Dropout(0.2)(x)  # FIXED: Was 0.4


outputs = Dense(train_generator.num_classes, activation='softmax')(x)


model = Model(base_model.input, outputs)


# FIXED: Remove label_smoothing (can hurt accuracy on small datasets)
loss_fn = tf.keras.losses.CategoricalCrossentropy()


# FIXED: Higher learning rate for initial training (frozen base)
model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-2),  # FIXED: Was 1e-3
    loss=loss_fn,
    metrics=['accuracy']
)


model.summary()


# ============================================================
# STEP 5: CLASS WEIGHTS
# ============================================================


class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_generator.classes),
    y=train_generator.classes
)
class_weights = dict(enumerate(class_weights))


print("Class weights:", class_weights)


# ============================================================
# CALLBACKS
# ============================================================


early_stop = tf.keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=7,  # FIXED: Was 5, increase patience
    restore_best_weights=True,
    verbose=1
)


reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,  # FIXED: Was 0.2, less aggressive
    patience=3,
    min_lr=1e-7,
    verbose=1
)


checkpoint = tf.keras.callbacks.ModelCheckpoint(
    "/content/Best_Maize_Model.keras",
    monitor='val_accuracy',
    save_best_only=True,
    save_weights_only=False,
    verbose=1
)


# ============================================================
# STEP 6: INITIAL TRAINING (FEATURE EXTRACTION)
# ============================================================


print("\n" + "="*60)
print("PHASE 1: Training with FROZEN base model (Feature Extraction)")
print("="*60)


history = model.fit(
    train_generator,
    validation_data=test_generator,
    epochs=30,  # FIXED: Was 50, sufficient for feature extraction
    class_weight=class_weights,
    callbacks=[early_stop, reduce_lr, checkpoint]
)


# ============================================================
# STEP 7: FINE TUNING (UNFREEZE PART OF BASE MODEL)
# ============================================================


print("\n" + "="*60)
print("PHASE 2: Fine-tuning with UNFROZEN base model")
print("="*60)


# Unfreeze base model
base_model.trainable = True


# Freeze earlier layers, unfreeze later layers (where features are more specific)
# VGG16 has fewer layers than EfficientNet, so unfreeze last 10 layers
for layer in base_model.layers[:-10]:
    layer.trainable = False


# FIXED: Compile with VERY LOW learning rate for fine-tuning
model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-5),  # Very low LR for fine-tuning
    loss=loss_fn,
    metrics=['accuracy']
)


print("Fine-tuning layers:", sum([1 for layer in base_model.layers if layer.trainable]))


# FIXED: More epochs for fine-tuning (was 5, now 15)
history2 = model.fit(
    train_generator,
    validation_data=test_generator,
    epochs=15,  # FIXED: Was 5
    class_weight=class_weights,
    callbacks=[early_stop, reduce_lr, checkpoint]
)


# ============================================================
# STEP 8: TRAINING & VALIDATION GRAPHS
# ============================================================


def plot_training_history(h1, h2=None):
    acc      = h1.history['accuracy']
    val_acc  = h1.history['val_accuracy']
    loss     = h1.history['loss']
    val_loss = h1.history['val_loss']


    if h2 is not None:
        acc      += h2.history['accuracy']
        val_acc  += h2.history['val_accuracy']
        loss     += h2.history['loss']
        val_loss += h2.history['val_loss']


    epochs_range = range(1, len(acc) + 1)
    split_epoch  = len(h1.history['accuracy'])


    fig, axes = plt.subplots(1, 2, figsize=(16, 6))
    fig.suptitle('Training & Validation Metrics', fontsize=16, fontweight='bold')


    ax = axes[0]
    ax.plot(epochs_range, acc,     color='royalblue',  linewidth=2, label='Train Accuracy')
    ax.plot(epochs_range, val_acc, color='tomato',     linewidth=2, label='Val Accuracy',
            linestyle='--')


    if h2 is not None:
        ax.axvline(x=split_epoch, color='green', linestyle=':', linewidth=2,
                   label=f'Fine-Tune Start (Epoch {split_epoch})')


    ax.set_title('Accuracy', fontsize=13)
    ax.set_xlabel('Epoch')
    ax.set_ylabel('Accuracy')
    ax.legend()
    ax.grid(True, alpha=0.3)
    ax.set_ylim([0, 1.05])


    ax = axes[1]
    ax.plot(epochs_range, loss,     color='royalblue',  linewidth=2, label='Train Loss')
    ax.plot(epochs_range, val_loss, color='tomato',     linewidth=2, label='Val Loss',
            linestyle='--')


    if h2 is not None:
        ax.axvline(x=split_epoch, color='green', linestyle=':', linewidth=2,
                   label=f'Fine-Tune Start (Epoch {split_epoch})')


    ax.set_title('Loss', fontsize=13)
    ax.set_xlabel('Epoch')
    ax.set_ylabel('Loss')
    ax.legend()
    ax.grid(True, alpha=0.3)


    plt.tight_layout()
    plt.savefig('/content/training_history.png', dpi=150, bbox_inches='tight')
    plt.show()
    print("Graph saved to /content/training_history.png")


plot_training_history(history, history2)


# ============================================================
# STEP 9: EVALUATION
# ============================================================


print("\n" + "="*60)
print("FINAL MODEL EVALUATION")
print("="*60)


pred   = model.predict(test_generator)
y_pred = np.argmax(pred, axis=1)
y_true = test_generator.classes


print("\n========== EVALUATION METRICS ==========")
print(f"Accuracy  : {accuracy_score(y_true, y_pred):.4f} ({accuracy_score(y_true, y_pred)*100:.2f}%)")
print(f"Precision : {precision_score(y_true, y_pred, average='weighted'):.4f}")
print(f"Recall    : {recall_score(y_true, y_pred, average='weighted'):.4f}")
print(f"F1 Score  : {f1_score(y_true, y_pred, average='weighted'):.4f}")


print("\nClassification Report:")
print(classification_report(
    y_true,
    y_pred,
    target_names=list(train_generator.class_indices.keys())
))


cm = confusion_matrix(y_true, y_pred)
class_names = list(train_generator.class_indices.keys())


plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d',
            xticklabels=class_names,
            yticklabels=class_names,
            cmap='Blues')
plt.title("Confusion Matrix", fontsize=14, fontweight='bold')
plt.xlabel("Predicted", fontsize=12)
plt.ylabel("Actual", fontsize=12)
plt.tight_layout()
plt.savefig('/content/confusion_matrix.png', dpi=150, bbox_inches='tight')
plt.show()


# ============================================================
# STEP 10: SAVE MODEL
# ============================================================


model.save("/content/Maize_Final_Model.keras")
print("\n✅ Model saved to /content/Maize_Final_Model.keras")


# ============================================================
# STEP 11: SINGLE IMAGE PREDICTION + RECOMMENDATION
# ============================================================


print("\n" + "="*60)
print("SINGLE IMAGE PREDICTION")
print("="*60)
print("Please upload an image to predict:")


uploaded_img = files.upload()
img_path = list(uploaded_img.keys())[0]


img       = load_img(img_path, target_size=(224, 224))
img_array = img_to_array(img)


original_image = img_array.copy().astype(np.uint8)
img_array = preprocess_input(img_array)
img_array = np.expand_dims(img_array, axis=0)


pred = model.predict(img_array)
cls  = np.argmax(pred)


predicted_disease = class_names[cls]
confidence        = np.max(pred) * 100


print("\n===================================")
print(f"Predicted Disease : {predicted_disease}")
print(f"Confidence Score  : {confidence:.2f}%")
print("===================================")


plt.figure(figsize=(6, 6))
plt.imshow(load_img(img_path))
plt.title(f"{predicted_disease}  ({confidence:.2f}%)", fontsize=14)
plt.axis("off")
plt.tight_layout()
plt.show()


print("\n--- Class Probabilities ---")
for i, name in enumerate(class_names):
    bar = '█' * int(pred[0][i] * 40)
    print(f"{name:<20}: {pred[0][i]*100:6.2f}%  {bar}")


# ============================================================
# ENHANCED RECOMMENDATION SYSTEM
# ============================================================


temperature = float(input("\nEnter Temperature (°C): "))
humidity    = float(input("Enter Humidity (%)    : "))


def get_risk_level(temp, hum, disease):
    if disease == "Anthracnose":
        if temp > 25 and hum > 80:
            return "🔴 HIGH RISK – Warm and humid conditions favor Anthracnose."
        elif hum > 70:
            return "🟡 MODERATE RISK – Monitor closely."
        else:
            return "🟢 LOW RISK – Conditions are less favorable."

    elif disease == "Bacterial Canker":
        if temp > 24 and hum > 75:
            return "🔴 HIGH RISK – Warm, wet weather favors bacterial spread."
        elif hum > 60:
            return "🟡 MODERATE RISK – Watch for symptoms."
        else:
            return "🟢 LOW RISK – Lower spread risk."

    elif disease == "Cutting Weevil":
        if temp > 20 and hum > 60:
            return "🟡 MODERATE RISK – Pest activity may increase."
        else:
            return "🟢 LOW RISK – Current conditions are less favorable."

    elif disease == "Die Back":
        if temp > 28 and hum > 70:
            return "🔴 HIGH RISK – Stress and humidity may worsen die back."
        elif hum > 60:
            return "🟡 MODERATE RISK – Monitor branch and shoot symptoms."
        else:
            return "🟢 LOW RISK – Reduced disease pressure."

    elif disease == "Gall Midge":
        if temp > 26 and hum > 65:
            return "🟡 MODERATE RISK – Conditions may support infestation."
        else:
            return "🟢 LOW RISK – Lower pest pressure."

    elif disease == "Powdery Mildew":
        if 20 <= temp <= 30 and hum > 60:
            return "🔴 HIGH RISK – Favorable conditions for powdery mildew."
        elif hum > 50:
            return "🟡 MODERATE RISK – Monitor leaf surfaces."
        else:
            return "🟢 LOW RISK – Less favorable for spread."

    elif disease == "Sooty Mould":
        if hum > 70:
            return "🟡 MODERATE RISK – High humidity can support sooty mould."
        else:
            return "🟢 LOW RISK – Current conditions are less favorable."

    elif disease == "Healthy":
        return "🟢 NO DISEASE DETECTED – Maintain routine management."

    else:
        return "⚠ UNKNOWN CONDITION"

def disease_recommendation(disease, temp, hum):
    risk = get_risk_level(temp, hum, disease)

    BASE = f"""
Temperature : {temp} °C
Humidity    : {hum} %
Risk Level  : {risk}
"""

    if disease == "Anthracnose":
        return f"""
==================================================
  DISEASE DETECTED : ANTHRACNOSE
==================================================
{BASE}
Recommended Actions:
  ✓ Remove and destroy infected leaves, twigs, and fruits.
  ✓ Apply copper-based or recommended fungicide early.
  ✓ Improve air circulation by pruning dense canopy.
  ✓ Avoid overhead irrigation.
  ✓ Monitor new flushes and fruits regularly.
"""

    elif disease == "Bacterial Canker":
        return f"""
==================================================
  DISEASE DETECTED : BACTERIAL CANKER
==================================================
{BASE}
Recommended Actions:
  ✓ Prune infected branches and destroy them.
  ✓ Disinfect pruning tools after use.
  ✓ Avoid wounds during wet weather.
  ✓ Use approved copper bactericides if recommended locally.
"""

    elif disease == "Cutting Weevil":
        return f"""
==================================================
  DISEASE DETECTED : CUTTING WEEVIL
==================================================
{BASE}
Recommended Actions:
  ✓ Remove infested plant parts and nearby weeds.
  ✓ Use pheromone or light traps if available.
  ✓ Apply recommended insecticide under expert guidance.
  ✓ Monitor trunks, shoots, and new growth.
"""

    elif disease == "Die Back":
        return f"""
==================================================
  DISEASE DETECTED : DIE BACK
==================================================
{BASE}
Recommended Actions:
  ✓ Prune affected shoots well below infected tissue.
  ✓ Improve drainage and reduce plant stress.
  ✓ Apply fungicide if fungal die back is confirmed.
  ✓ Avoid waterlogging and nutrient imbalance.
"""

    elif disease == "Gall Midge":
        return f"""
==================================================
  DISEASE DETECTED : GALL MIDGE
==================================================
{BASE}
Recommended Actions:
  ✓ Remove and destroy affected shoots.
  ✓ Use recommended insect control measures.
  ✓ Monitor young flushes closely.
  ✓ Maintain orchard sanitation.
"""

    elif disease == "Powdery Mildew":
        return f"""
==================================================
  DISEASE DETECTED : POWDERY MILDEW
==================================================
{BASE}
Recommended Actions:
  ✓ Apply sulfur-based or recommended fungicides early.
  ✓ Improve sunlight penetration and airflow.
  ✓ Avoid excess nitrogen fertilizer.
  ✓ Remove heavily infected foliage.
"""

    elif disease == "Sooty Mould":
        return f"""
==================================================
  DISEASE DETECTED : SOOTY MOULD
==================================================
{BASE}
Recommended Actions:
  ✓ Control honeydew-producing insects like aphids or scales.
  ✓ Wash affected leaves if appropriate.
  ✓ Improve canopy ventilation.
  ✓ Remove severely affected debris.
"""

    elif disease == "Healthy":
        return f"""
==================================================
  ✅ HEALTHY MANGO LEAF DETECTED
==================================================
{BASE}
Routine Management:
  ✓ Continue normal orchard care.
  ✓ Monitor for pests and early disease signs.
  ✓ Maintain pruning, sanitation, and drainage.
"""

    else:
        return f"""
==================================================
  ⚠ UNKNOWN / UNRECOGNISED CONDITION
==================================================
{BASE}
Recommended Actions:
  ✓ Consult a plant health specialist.
  ✓ Capture clearer images.
  ✓ Inspect additional leaves and shoots.
"""


print(disease_recommendation(predicted_disease, temperature, humidity))


# ============================================================
# PROBABILITY BAR CHART
# ============================================================


probabilities = pred[0] * 100
colors = ['#2ecc71' if c == predicted_disease else '#3498db' for c in class_names]


plt.figure(figsize=(9, 5))
bars = plt.bar(class_names, probabilities, color=colors, edgecolor='black', linewidth=0.7)
plt.title("Disease Probability Distribution", fontsize=14, fontweight='bold')
plt.ylabel("Probability (%)")
plt.xlabel("Disease Class")
plt.ylim(0, 115)


for bar, val in zip(bars, probabilities):
    plt.text(bar.get_x() + bar.get_width() / 2,
             bar.get_height() + 1.5,
             f"{val:.1f}%",
             ha='center', va='bottom', fontsize=10)


plt.tight_layout()
plt.savefig('/content/probability_chart.png', dpi=150, bbox_inches='tight')
plt.show()


# ============================================================
# STEP 12: GRAD-CAM++ VISUALISATION
# ============================================================


img_gc       = load_img(img_path, target_size=(224, 224))
img_gc_array = img_to_array(img_gc)
img_gc_proc  = preprocess_input(img_gc_array.copy())
img_gc_proc  = np.expand_dims(img_gc_proc, axis=0)


score   = CategoricalScore([int(cls)])
gradcam = GradcamPlusPlus(model, clone=True)
cam     = gradcam(score, img_gc_proc)


heatmap = cam[0]


fig, axes = plt.subplots(1, 3, figsize=(15, 5))


axes[0].imshow(img_gc)
axes[0].set_title("Original Image")
axes[0].axis("off")


axes[1].imshow(heatmap, cmap='jet')
axes[1].set_title("Grad-CAM++ Heatmap")
axes[1].axis("off")


axes[2].imshow(img_gc)
axes[2].imshow(heatmap, cmap='jet', alpha=0.45)
axes[2].set_title(f"Overlay – {predicted_disease}")
axes[2].axis("off")


plt.suptitle("Grad-CAM++ Explainability", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('/content/gradcam_output.png', dpi=150, bbox_inches='tight')
plt.show()


# ============================================================
# FINAL SUMMARY
# ============================================================


print("\n" + "="*60)
print("✅ DONE – HIGH ACCURACY MAIZE DISEASE DETECTION MODEL (VGG16) READY 🚀")
print("="*60)
print("\nSaved files:")
print("  ✓ /content/Best_Maize_Model.keras  (Best model)")
print("  ✓ /content/Maize_Final_Model.keras (Final model)")
print("  ✓ /content/training_history.png")
print("  ✓ /content/confusion_matrix.png")
print("  ✓ /content/probability_chart.png")
print("  ✓ /content/gradcam_output.png")


print("\n" + "="*60)
print("KEY CHANGES FOR VGG16:")
print("="*60)
print("1. ✅ Changed from EfficientNetB0 to VGG16")
print("2. ✅ Updated import: tensorflow.keras.applications.vgg16")
print("3. ✅ Updated preprocess_input from vgg16")
print("4. ✅ Adjusted fine-tuning layers (unfreeze last 10 instead of 30)")
print("5. ✅ VGG16 has simpler architecture - fewer layers to unfreeze")
print("="*60)
```
